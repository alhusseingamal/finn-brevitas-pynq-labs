```python
import json
import time
import numpy as np
from driver import FINNDMAOverlay
from dataset import GTSRBDataset

# file paths
BITFILE_PATH = "./bitfiles/task4.bit"
SETTINGS_FILE = "./settings_task4.json"
BATCH_SIZE = 1000

print("Initializing FINN Overlay...")
with open(SETTINGS_FILE, "r") as f:
    driver_settings = json.load(f)["driver_information"]

accel = FINNDMAOverlay(
    BITFILE_PATH,
    batch_size=BATCH_SIZE,
    **driver_settings
)

expected_shape = accel.ishape_normal()
print(f"Hardware expects input shape (per batch): {expected_shape}")

# load/download dataset
print("Downloading/Loading GTSRB Test Dataset...")
dataset = GTSRBDataset(data_dir="./data", split="test", img_size=12, download=True) # model is trained on 12x12 images

total_images = len(dataset)
n_batches = total_images // BATCH_SIZE
images_to_evaluate = n_batches * BATCH_SIZE

if images_to_evaluate == 0:
    raise ValueError("Not enough images for a single batch of 1000!")

print(f"Successfully loaded {total_images} test images.")

# allocate memory based on FINN expected shape: (n_batches, 1000, 12, 12, 3)
input_data = np.zeros((n_batches, BATCH_SIZE, 12, 12, 3), dtype=np.uint8)
labels = np.zeros((n_batches, BATCH_SIZE), dtype=np.int64)

for b in range(n_batches):
    for i in range(BATCH_SIZE):
        idx = b * BATCH_SIZE + i
        img_np, label = dataset[idx]
        input_data[b, i] = img_np
        labels[b, i] = label

print(f"Packed {images_to_evaluate} images into {n_batches} batches...")

# DMA needs contiguous memory chunks
input_data = np.ascontiguousarray(input_data)

# Batched Inference
print("Starting hardware inference via DMA...")
predictions = np.zeros((n_batches, BATCH_SIZE), dtype=np.int64)

start_time = time.time()

for b in range(n_batches):
    output = accel.execute(input_data[b])
    predictions[b] = output.flatten()

end_time = time.time()

# Metrics
total_time = end_time - start_time
fps = images_to_evaluate / total_time
accuracy = np.mean(predictions == labels) * 100

print("\n========================================")
print("          EVALUATION RESULTS")
print("\n========================================")
print(f"{images_to_evaluate} Images Evaluated in batches of {BATCH_SIZE}")
print(f"Total Latency:          {total_time:.4f} seconds")
print(f"Hardware Throughput:    {fps:.2f} FPS")
print(f"Model Accuracy:         {accuracy:.2f}%")
print("\n========================================")
```
